```python
#レジームフィルターを適用
import numpy as np
import pandas as pd
df_final = df.copy()
# (df_finalが作成済みであると仮定)
print("\n--- レジームフィルターを適用し、学習データを最終化します ---")

TARGET_VARIABLE = "Close"
# --- 1. レジームの定義 ---
# ADXが25以上、かつ価格が75日移動平均線の上なら「上昇トレンド」
# ADXが25以上、かつ価格が75日移動平均線の下なら「下降トレンド」
# それ以外は「レンジ相場」とする

conditions = [
    (df_final['Adx_14'] >= 25) & (df_final[TARGET_VARIABLE] > df_final['Sma_75']),
    (df_final['Adx_14'] >= 25) & (df_final[TARGET_VARIABLE] < df_final['Sma_75']),
]
choices = ['上昇トレンド', '下降トレンド']
df_final['Market_Regime'] = np.select(conditions, choices, default='レンジ相場')

# --- 2. レジームに応じた特徴量の選択 ---

# まず、全てのモデルで共通して使う基本的な特徴量を定義
base_features = [
    # ここに、どのレジームでも重要と考える基本的な特徴量をリストアップ
    '終値前日比', 'Volume_Shape', 'Rsi_14', 'Rolling_Std_5D' 
]

# トレンド相場で特に重要になる指標
trend_features = [
    'Adx_14', 'Macd_12_26_9', 'Sma_5', 'Sma_25', 
    'Psarl_0.02_0.2', 
    'Psars_0.02_0.2' 
]

# レンジ相場で特に重要になる指標
range_features = [
    'Stochk_14_3_3', 'Bbp_20_2.0', 'Bb_Width'
]

# --- 3. 最終的な学習データ(X)とターゲット(y)の作成 ---

# ターゲット変数 (例: 翌日の終値が上がるか下がるか)
# (これはあなたのLGPHモデルの目的に合わせて定義してください)
df_final['target'] = (df_final[TARGET_VARIABLE].shift(-1) > df_final[TARGET_VARIABLE]).astype(int)
df_final.dropna(inplace=True)

# 最終的な特徴量リストとデータフレームを準備
X_final = pd.DataFrame()
y_final = pd.Series(dtype=int)

# レジームごとにデータを分割し、対応する特徴量セットで学習データを構築
for regime in ['上昇トレンド', '下降トレンド', 'レンジ相場']:
    
    regime_df = df_final[df_final['Market_Regime'] == regime].copy()
    
    if regime == '上昇トレンド' or regime == '下降トレンド':
        selected_features = base_features + trend_features
    else: # レンジ相場
        selected_features = base_features + range_features
        
    # 選択された特徴量のみを持つデータフレームを作成
    X_regime = regime_df[selected_features]
    y_regime = regime_df['target']
    
    # 全体を結合していく
    X_final = pd.concat([X_final, X_regime])
    y_final = pd.concat([y_final, y_regime])
    
    X_final.fillna(0, inplace=True)

# インデックスをソートして時系列を正しく保つ
X_final.sort_index(inplace=True)
y_final.sort_index(inplace=True)

print("\nレジームフィルターの適用完了。")
print(f"最終的な特徴量の数: {len(X_final.columns)}個")
print("--- 学習データの準備が全て完了しました ---")
print(X_final.head())
print(y_final.head())
```

```python
#LGPHの構築と保存
import os
import pandas as pd
import numpy as np
import lightgbm as lgb
from prophet import Prophet
import japanize_matplotlib
import joblib # モデルの保存とロードに使う
from sklearn.metrics import mean_absolute_error, mean_squared_error
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['font.family'] = 'Yu Gothic'
# === ステップ3：レジームフィルター適用と、AIの全力再学習のためのデータ準備 ===
# --- ▼▼▼ このブロックを追加 ▼▼▼ ---
prophet_learn_df = pd.DataFrame({
    'ds': df['ds'],
    'y': df[TARGET_VARIABLE] # ベースのトレンドを学習
})
final_df = df.copy()
# === 準備2：Prophetによる外れ値の再定義と全特徴量の構築 (完全版) ===
print("\n--- 準備2: Prophetによる外れ値の再定義と、全特徴量の構築 ---")
prophet_learn_df = pd.DataFrame({'ds': df['ds'], 'y': df[TARGET_VARIABLE]})
model = Prophet(interval_width=0.05)
model.fit(prophet_learn_df)
forecast = model.predict(prophet_learn_df[['ds']])
final_df = pd.merge(df, forecast[['ds', 'yhat', 'yhat_upper', 'yhat_lower']], on='ds', how='inner')

# 仕上げ
final_df['y'] = final_df[TARGET_VARIABLE]
conditions = [final_df['y'] > final_df['yhat_upper'], final_df['y'] < final_df['yhat_lower']]
choices = [final_df['y'] - final_df['yhat_upper'], final_df['y'] - final_df['yhat_lower']]
final_df['outlier_error'] = np.select(conditions, choices, default=0.0)
final_df.dropna(inplace=True)
final_df.reset_index(drop=True, inplace=True)
print("外れ値の再定義と全特徴量の構築 完了。\n")

conditions_outlier = [
    final_df[TARGET_VARIABLE] > final_df['yhat_upper'],
    final_df[TARGET_VARIABLE] < final_df['yhat_lower']
]
choices_outlier = [
    final_df[TARGET_VARIABLE] - final_df['yhat_upper'],
    final_df[TARGET_VARIABLE] - final_df['yhat_lower']
]
final_df['outlier_error'] = np.select(conditions_outlier, choices_outlier, default=0.0)
print("'outlier_error'カラムを計算しました。")
# --- ▲▲▲ 追加ここまで ▲▲▲ ---


# --- 3A: レジームの定義 ---
conditions = [
    (final_df['Adx_14'] > 25) & (final_df[TARGET_VARIABLE] > final_df['Sma_75']),
    (final_df['Adx_14'] > 25) & (final_df[TARGET_VARIABLE] < final_df['Sma_75']),
]
choices = ['上昇トレンド', '下降トレンド']
final_df['Market_Regime'] = np.select(conditions, choices, default='レンジ相場')

# --- 3B: 3つの目的変数の作成 ---
# 'outlier_error'が存在するので、この部分はエラーなく実行される
final_df['is_outlier'] = (final_df['outlier_error'] != 0).astype(int)
final_df['error_sign'] = np.sign(final_df['outlier_error']).astype(int)
final_df['log_abs_error'] = np.log1p(abs(final_df['outlier_error']))
print("3つの目的変数（is_outlier, error_sign, log_abs_error）を作成しました。")

# --- 3C: 特徴量リストの定義 ---
# 除外リストからTARGET_VARIABLEを削除 (レジームフィルターで使うため)
cols_to_exclude = [
    'ds', 'y', 'yhat', 'yhat_upper', 'yhat_lower', 'outlier_error',
    'is_outlier', 'error_sign', 'log_abs_error', 'Market_Regime', 'target'
]
base_feature_candidates = [col for col in final_df.columns if col not in cols_to_exclude and TARGET_VARIABLE not in col]
# SHAP等で重要度が高かったものを厳選
base_features = ['終値前日比', 'Volume_Shape', 'Rsi_14', 'Rolling_Std_5D']
trend_features = ['Adx_14', 'Macd_12_26_9', 'Sma_5', 'Sma_25', 'Psarl_0.02_0.2', 'Psars_0.02_0.2']
range_features = ['Stochk_14_3_3', 'Bbp_20_2.0', 'Bb_Width']

# --- 3D: レジームフィルターを適用した最終学習データ(X_final, y_..._final)の作成 ---
X_final = pd.DataFrame()
y_class_final = pd.Series(dtype=int)
y_reg_log_final = pd.Series(dtype=float)
y_sign_final = pd.Series(dtype=int)

for regime in ['上昇トレンド', '下降トレンド', 'レンジ相場']:
    regime_df = final_df[final_df['Market_Regime'] == regime].copy()
    
    if regime == '上昇トレンド' or regime == '下降トレンド':
        selected_features = base_features + trend_features
    else: # レンジ相場
        selected_features = base_features + range_features
        
    X_regime = regime_df[selected_features]
    
    X_final = pd.concat([X_final, X_regime])
    y_class_final = pd.concat([y_class_final, regime_df['is_outlier']])
    y_reg_log_final = pd.concat([y_reg_log_final, regime_df['log_abs_error']])
    y_sign_final = pd.concat([y_sign_final, regime_df['error_sign']])

# NaNを0で埋め、時系列を正しく保つ
X_final.fillna(0, inplace=True)
X_final.sort_index(inplace=True)
y_class_final.sort_index(inplace=True)
y_reg_log_final.sort_index(inplace=True)
y_sign_final.sort_index(inplace=True)

print("レジームフィルターの適用完了。")


# === ステップ4：LGPH専門家チームの「全力」学習と、知識の保存 (X_final版) ===
print("\n--- ステップ4: LGPH専門家チームを学習させ、その知識を保存します ---")

MODELS_DIR = "lgph_supervised_models"
os.makedirs(MODELS_DIR, exist_ok=True)

# --- モデル1：門番 (分類モデル) の学習と保存 ---
print("【モデル1】 分類モデル（門番）を学習中...")
# y_class_finalが空でない、かつユニークな値が2つ以上あることを確認
if not y_class_final.empty and y_class_final.nunique() >= 2:
    scale_pos_weight = y_class_final.value_counts().get(0, 0) / y_class_final.value_counts().get(1, 1)
    lgbm_clf = lgb.LGBMClassifier(
        objective='binary', scale_pos_weight=scale_pos_weight, random_state=42, verbosity=-1,
        num_leaves=40, max_depth=7, learning_rate=0.05, n_estimators=500
    )
    lgbm_clf.fit(X_final, y_class_final)
    joblib.dump(lgbm_clf, os.path.join(MODELS_DIR, f'{synbol}_legeam_lgbm_clf.pkl'))
    print(f"門番の知識を '{synbol}_legeam_lgbm_clf.pkl' として保存しました。")
else:
    print("門番：学習データが存在しない、または1種類しかないためスキップしました。")


# --- モデル2：大きさの専門家チーム (Quantile Regression) の学習と保存 ---
print("\n【モデル2】 大きさの専門家チーム (Quantile Regression) を学習中...")
is_outlier_final = (y_class_final == 1)
X_reg_final = X_final[is_outlier_final]
y_reg_log_final_only = y_reg_log_final[is_outlier_final]

lgbm_reg_quantile_models = {}
if not y_reg_log_final_only.empty:
    quantiles = {"lower": 0.10, "median": 0.50, "upper": 0.90}
    for name, alpha in quantiles.items():
        params = {
            'objective': 'quantile', 'alpha': alpha, 'random_state': 42, 'verbosity': -1,
            'num_leaves': 50, 'max_depth': 10, 'learning_rate': 0.02, 'n_estimators': 1000
        }
        model = lgb.LGBMRegressor(**params)
        model.fit(X_reg_final, y_reg_log_final_only)
        lgbm_reg_quantile_models[name] = model
    
    joblib.dump(lgbm_reg_quantile_models, os.path.join(MODELS_DIR, f'{synbol}_legeam_lgbm_reg_quantiles.pkl'))
    print(f"大きさの専門家チームの知識を '{synbol}_legeam_lgbm_reg_quantiles.pkl' として保存しました。")
else:
    print("大きさ専門家：学習データが存在せず、スキップしました。")


# --- モデル3：方向の専門家 (分類モデル) の学習と保存 ---
print("\n【モデル3】 符号予測モデル（方向の専門家）を学習中...")
X_sign_final = X_final[is_outlier_final]
y_sign_final_only = y_sign_final[is_outlier_final]

if not y_sign_final_only.empty:
    X_sign_final = X_sign_final[y_sign_final_only != 0]
    y_sign_final_only = y_sign_final_only[y_sign_final_only != 0]

if not y_sign_final_only.empty and y_sign_final_only.nunique() >= 2:
    lgbm_sign_clf = lgb.LGBMClassifier(
        objective='binary', random_state=42, verbosity=-1,
        num_leaves=31, max_depth=5, learning_rate=0.1, n_estimators=300
    )
    lgbm_sign_clf.fit(X_sign_final, y_sign_final_only)
    joblib.dump(lgbm_sign_clf, os.path.join(MODELS_DIR, f'{synbol}_legeam_lgbm_sign_clf.pkl'))
    print(f"方向の専門家の知識を '{synbol}_legeam_lgbm_sign_clf.pkl' として保存しました。")
else:
    print("方向の専門家：学習データが存在しない（または1種類しかない）ため、スキップしました。")

print(f'\n--- {synbol} の全学習保存完了！ ---')
```

```python
from scipy.stats import skew, kurtosis
import shap

initial_balance=10000
commission_rate=0.00075
slippage=0.00050
stop_loss_pct=-0.05

def run_backtest(risk_tolerance, X_test_final, prophet_base_predictions_test, actual_prices_test, dates_test,
                 clf_model, reg_models, sign_model,
                 initial_balance=10000, 
                 commission_rate=0.00075, 
                 slippage=0.00050, 
                 stop_loss_pct=-0.05):
    """
    X_finalのロジック（レジームフィルター）を組み込んだ、実践用のバックテスト関数。
    """
    # === 初期設定 (変更なし) ===
    equity = initial_balance
    working_capital = initial_balance
    position_ratio = 0.0
    amount_held = 0.0
    average_entry_price = 0.0
    history = []
    total_cost = commission_rate + slippage

    print("X_final版バックテストを開始します...")

    for i in range(len(X_test_final)):
        # ▼▼▼ ここからがX_final版のロジック ▼▼▼
        # X_test_finalから、その日の特徴量セットを直接取得
        current_features = X_test_final.iloc[[i]]
        # ▲▲▲ ここまで ▲▲▲
        
        current_price = actual_prices_test[i]
        current_date = dates_test[i]
        daily_trade_value = 0

        # --- 1. 日々の損益計算 (変更なし) ---
        if i > 0 and amount_held != 0:
            pnl = amount_held * (current_price - actual_prices_test[i-1])
            equity += pnl

        if equity <= 0:
            print(f"{current_date}に資産が枯渇しました。")
            break

        # --- 2. 目標ポジション比率の決定 (変更なし) ---
        target_position_ratio = position_ratio
        unrealized_pnl_pct = 0
        if amount_held > 0: unrealized_pnl_pct = (current_price / average_entry_price) - 1
        elif amount_held < 0: unrealized_pnl_pct = (average_entry_price / current_price) - 1
        if unrealized_pnl_pct < stop_loss_pct:
            target_position_ratio = 0.0
        
        if target_position_ratio == position_ratio:
            if i == len(X_test_final) - 1:
                prediction_final = prophet_base_predictions_test[i]
                predicted_change_pct = (prediction_final / current_price) - 1
            else:
                yhat_tomorrow = prophet_base_predictions_test[i+1]
                prediction_tomorrow = yhat_tomorrow
                
                # ★★★ X_final版の特徴量(current_features)を使って予測 ★★★
                class_pred = clf_model.predict(current_features)[0]
                if class_pred == 1:
                    log_abs_error_pred = reg_models['median'].predict(current_features)[0]
                    abs_error_pred = np.expm1(log_abs_error_pred)
                    sign_pred = sign_model.predict(current_features)[0]
                    prediction_tomorrow += abs_error_pred * sign_pred
                
                predicted_change_pct = (prediction_tomorrow / current_price) - 1

            confidence_score = np.clip(abs(predicted_change_pct) / risk_tolerance, 0, 1) if risk_tolerance > 0 else 1
            target_direction = np.sign(predicted_change_pct)
            target_position_ratio = target_direction * confidence_score

        # --- 3. 取引の実行 (変更なし) ---
        position_ratio_change = target_position_ratio - position_ratio
        if position_ratio_change != 0:
            # ( ... 取引実行のロジックは変更ありません ... )
            equity_before_trade = equity
            trade_notional_value = equity_before_trade * abs(position_ratio_change)
            cost = trade_notional_value * total_cost
            equity -= cost
            working_capital -= cost
            trade_amount = trade_notional_value / current_price * np.sign(position_ratio_change)
            if np.sign(position_ratio) != np.sign(target_position_ratio) and position_ratio != 0:
                working_capital += abs(amount_held) * current_price
                opening_value = equity_before_trade * abs(target_position_ratio)
                working_capital -= opening_value
                amount_held = opening_value / current_price * np.sign(target_position_ratio)
                average_entry_price = current_price if target_position_ratio != 0 else 0
            elif abs(target_position_ratio) > abs(position_ratio):
                if amount_held != 0:
                    average_entry_price = ((average_entry_price * abs(amount_held)) + trade_notional_value) / (abs(amount_held) + abs(trade_amount))
                else: average_entry_price = current_price
                working_capital -= trade_notional_value
                amount_held += trade_amount
            else:
                working_capital += trade_notional_value
                amount_held += trade_amount
                if abs(amount_held) < 1e-9: amount_held, average_entry_price = 0, 0
            daily_trade_value = trade_notional_value

        # --- 4. 状態の記録 & 最終決済 (変更なし) ---
        position_ratio = target_position_ratio
        history.append({
            'ds': current_date, 'equity': equity, 'working_capital': working_capital,
            'position_value': abs(amount_held) * current_price, 'action': position_ratio,
            'amount_held': amount_held, 'actual_price': current_price,
            'average_entry_price': average_entry_price, 'trade_value': daily_trade_value,
            'risk_tolerance': risk_tolerance
        })
    
    if history and history[-1]['amount_held'] != 0:
        # ( ... 最終決済のロジックは変更ありません ... )
        last_record = history[-1]
        final_price = last_record['actual_price']
        amount_to_close = last_record['amount_held']
        final_liquidation_value = abs(amount_to_close) * final_price
        final_cost = final_liquidation_value * total_cost
        last_record['equity'] -= final_cost
        last_record['working_capital'] += final_liquidation_value - final_cost
        last_record['position_value'] = 0
        last_record['trade_value'] += final_liquidation_value
        
    return history

def objective_function(risk_tolerance, clf_model, reg_models, sign_model):
    """
    指定されたリスク重視度でバックテストを実行し、総合スコアだけを計算する。
    SHAPなどの重い処理はここで行わない。
    """
    print(f"--- リスク重視度: {risk_tolerance:.4f} でバックテストを実行中 ---")
    
    history = run_backtest(
        risk_tolerance, X_test_final, prophet_base_predictions_test, 
        actual_prices_test, dates_test,
        clf_model=clf_model,     # ★★★ モデルを渡す ★★★
        reg_models=reg_models,   # ★★★ モデルを渡す ★★★
        sign_model=sign_model    # ★★★ モデルを渡す ★★★
    )
    
    # --- パフォーマンス指標と総合スコアの計算 ---
    if not history or len(history) < 2: return -999
    
    plot_df = pd.DataFrame(history)
    # (ここから下のスコア計算部分は、あなたのコードと同じで問題ありません)
    # ... (Modified Sharpe Ratio, Calmarレシオ, 総合スコアの計算) ...
    
    equity_series = plot_df['equity']
    daily_returns = equity_series.pct_change().dropna()
    if len(daily_returns) < 2: return -999

    mean_return = daily_returns.mean()
    std_dev = daily_returns.std()
    if std_dev == 0:
        modified_sharpe_ratio = 0
    else:
        skewness = skew(daily_returns)
        kurt = kurtosis(daily_returns)
        sharpe_ratio_daily = mean_return / std_dev
        msr_adj = 1 + (skewness / 6) * sharpe_ratio_daily - ((kurt) / 24) * (sharpe_ratio_daily**2)
        modified_sharpe_ratio = (sharpe_ratio_daily * msr_adj) * np.sqrt(252)

    initial_balance_run = history[0]['equity']
    final_balance = equity_series.iloc[-1]
    total_return = (final_balance / initial_balance_run) - 1
    days = (plot_df['ds'].iloc[-1] - plot_df['ds'].iloc[0]).days
    annual_return = (1 + total_return) ** (365.0 / days) - 1 if days > 0 else 0
    cumulative_max = equity_series.cummax()
    max_drawdown = (equity_series - cumulative_max).min() / cumulative_max.max()
    calmar_ratio = annual_return / abs(max_drawdown) if max_drawdown != 0 else 100
        
    score = modified_sharpe_ratio * calmar_ratio
    
    print(f"結果: ModSR={float(modified_sharpe_ratio):.2f}, Calmar={float(calmar_ratio):.2f}, ★総合スコア={float(score):.2f}★")
    
    return score if pd.notna(score) and score > 0 else 0
# === 2. 黄金分割探索の実装 (変更なし) ===

def golden_section_search(f, a, b, tol=1e-3, max_iter=20):
    gr = (np.sqrt(5) + 1) / 2
    c = b - (b - a) / gr
    d = a + (b - a) / gr
    fc = f(c)
    fd = f(d)
    
    iterations = 0
    while abs(c - d) > tol and iterations < max_iter:
        if fc > fd:
            b, d, fd = d, c, fc
            c = b - (b - a) / gr
            fc = f(c)
        else:
            a, c, fc = c, d, fd
            d = a + (b - a) / gr
            fd = f(d)
        
        iterations += 1
        print(f"\n反復回数: {iterations}, 現在の最適範囲: [{a:.4f}, {b:.4f}]")

    optimal_x = (a + b) / 2
    max_score = f(optimal_x)
    
    return optimal_x, max_score

# === 3. 最適化の実行 ===

# リスク重視度の探索範囲を定義 (例: 1% ～ 30%)
lower_bound = 0.01
upper_bound = 0.30

print("\n" + "="*50)
print("黄金分割探索による最適なリスク重視度の探索を開始します。")
print(f"探索範囲: [{lower_bound:.2%}, {upper_bound:.2%}]")
print("="*50 + "\n")

# ▼▼▼ このlambda式の定義を追加・確認 ▼▼▼
objective_with_models = lambda rt: objective_function(
    rt, clf_model, reg_models, sign_model
)

# 黄金分割探索を実行
optimal_risk_tolerance, max_score = golden_section_search(objective_with_models, lower_bound, upper_bound)

# === 最適化結果で最終バックテストを実行し、詳細な指標を取得 ===
print("\n--- 最適化されたリスク重視度で最終指標を計算中... ---")
final_history = run_backtest(
    optimal_risk_tolerance,
    X_test_final,
    prophet_base_predictions_test,
    actual_prices_test,
    dates_test,
    clf_model=clf_model,
    reg_models=reg_models,
    sign_model=sign_model
)

# 最終指標の計算
if final_history:
    plot_df = pd.DataFrame(final_history)
    final_equity = plot_df['equity']
    final_returns = final_equity.pct_change().dropna()

    final_sharpe = (final_returns.mean() / final_returns.std()) * np.sqrt(252) if final_returns.std() > 0 else 0
    
    final_cumulative_max = final_equity.cummax()
    final_drawdown_series = (final_equity - final_cumulative_max) / final_cumulative_max
    final_max_drawdown = final_drawdown_series.min()
    final_initial_balance = final_history[0]['equity']
    final_final_balance = final_history[-1]['equity']
    final_total_return = (final_final_balance / final_initial_balance) - 1
    
    final_pnl = final_equity.diff().dropna()
    final_gross_profit = final_pnl[final_pnl > 0].sum()
    final_gross_loss = abs(final_pnl[final_pnl < 0].sum())
    final_profit_factor = final_gross_profit / final_gross_loss if final_gross_loss > 0 else 0
else:
    final_sharpe, final_max_drawdown, final_profit_factor = 0, 0, 0

# === 最適化完了レポート ===
print("\n" + "="*50)
print("          最適化完了")
print("="*50)
# 基本パフォーマンスを表示
print(f"\n【教師ありレジームフィルタLGPH {synbol}版バックテスト レポート】")
print(f"初期資産: {final_initial_balance:,.2f}")
print(f"最終資産: {final_final_balance:,.2f}")
print(f"トータルリターン: {final_total_return:.2%}")
print("-" * 20) # 区切り線
print(f"★ 最適なリスク重視度: {optimal_risk_tolerance:.4f} (つまり {optimal_risk_tolerance:.2%})")
print(f"★ 最大総合スコア: {float(max_score):.2f}")
print("--- その時の詳細パフォーマンス ---")
print(f"シャープレシオ (年率): {final_sharpe:.2f}")
print(f"最大ドローダウン: {final_max_drawdown:.2%}")
print(f"プロフィットファクター: {final_profit_factor:.2f}")
```
